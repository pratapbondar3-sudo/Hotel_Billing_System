```python
!pip install qrcode
!pip install pillow
import pandas as pd
import qrcode
from PIL import Image
from IPython.display import display
import os


menu = {
    "Pizza": 250,
    "Burger": 120,
    "Pasta": 180,
    "Coffee": 80,
    "Tea": 40,
    "Sandwich": 100,
    "Cold Drink": 60,
    "Biryani": 220
}

print("========== HOTEL MENU ==========")

for item, price in menu.items():
    print(item, "= Rs", price)

customer_name = input("Enter Customer Name: ")
phone = input("Enter Mobile Number: ")

cart = []
total = 0

while True:

    item = input("Enter Food Item: ")

    if item not in menu:
        print("Item Not Available")
        continue

    qty = int(input("Enter Quantity: "))

    price = menu[item] * qty

    total = total + price

    cart.append([item, qty, price])

    more = input("Add More Items? (yes/no): ")

    if more.lower() == "no":
        break

print("\n========== BILL ==========")

for data in cart:
    print(data[0], "x", data[1], "= Rs", data[2])

print("\nSubtotal = Rs", total)

gst = total * 0.05

print("GST 5% = Rs", round(gst, 2))

final_total = total + gst

print("Final Total = Rs", round(final_total, 2))

upi_id = "pratapbondar06@ybl"

name = "Hotel Payment"

amount = str(round(final_total, 2))

upi_link = f"upi://pay?pa={upi_id}&pn={name}&am={amount}&cu=INR"

qr = qrcode.make(upi_link)

qr.save("phonepe_qr.png")

from IPython.display import display
display(qr)

print("\n==============================")
print("        HOTEL RECEIPT")
print("==============================")

print("Customer Name:", customer_name)
print("Mobile Number:", phone)

for data in cart:
    print(data[0], "x", data[1], "= Rs", data[2])

print("------------------------------")

print("Subtotal:", total)

print("GST 5%:", round(gst, 2))

print("Final Total:", round(final_total, 2))

print("Payment Mode: PhonePe")

print("==============================")

bill_data = {
    "Customer": [customer_name],
    "Phone": [phone],
    "Total": [round(final_total, 2)]
}

bill_df = pd.DataFrame(bill_data)

bill_df.to_csv(
    "bills.csv",
    mode='a',
    header=False,
    index=False
)

print("Bill Saved Successfully")

desktop_path = os.path.join(os.path.expanduser("~"), "Desktop")

bill_data = {
    "Customer": [customer_name],
    "Phone": [phone],
    "Total": [round(final_total, 2)]
}

bill_df = pd.DataFrame(bill_data)

bill_path = os.path.join(desktop_path, "bills.csv")

bill_df.to_csv(
    bill_path,
    mode='a',
    header=not os.path.exists(bill_path),
    index=False
)

print("Bill Saved Successfully on Desktop")
print("Bill File Location:", bill_path)
```
